##### Simple Perceptron example from scratch

```python
import pandas as pd

df = pd.read_csv("data/perceptron_toydata-truncated.txt", sep="\t")
df.head()
```

```python
X_train = df[["x1", "x2"]].values
y_train = df["label"].values
```

```python
X_train = torch.from_numpy(X_train)
y_train = torch.from_numpy(y_train)
```

```python
print(X_train.shape, y_train.shape)
```

```python
X_train = X_train.to(torch.float32)
y_train = y_train.to(torch.float32)
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
```

```python
# plot the data
plt.plot(
    X_train[y_train == 0, 0],
    X_train[y_train == 0, 1],
    marker="D",
    markersize=10,
    linestyle="",
    label="Class 0",
)

plt.plot(
    X_train[y_train == 1, 0],
    X_train[y_train == 1, 1],
    marker="^",
    markersize=13,
    linestyle="",
    label="Class 1",
)

plt.legend(loc=2)

plt.xlim([-5, 5])
plt.ylim([-5, 5])

plt.xlabel("Feature $x_1$", fontsize=12)
plt.ylabel("Feature $x_2$", fontsize=12)

plt.grid()
plt.show()
```

```python
class Perceptron:
    def __init__(self, num_features):
        self.num_features = num_features
        self.weights = torch.zeros(num_features)
        self.bias = torch.tensor(0.)

    def forward(self, x):
        weighted_sum_z = torch.dot(x, self.weights) + self.bias

        if weighted_sum_z > 0.:
            prediction = torch.tensor(1.)
        else:
            prediction = torch.tensor(0.)

        return prediction

    def update(self, x, true_y):
        prediction = self.forward(x)
        error = true_y - prediction

        # update
        self.bias += error
        self.weights += error * x

        return error
```

```python
ppn = Perceptron(num_features=2)
print("Model parameters:")
print("  Weights:", ppn.weights)
print("  Bias:", ppn.bias)
```

```python
x = torch.tensor([1.1, 2.1])
ppn.forward(x)
```

```python
ppn.update(x, true_y=1)
```

```python
def train(model, all_x, all_y, epochs):

    for epoch in range(epochs):
        error_count = 0

        for x, y in zip(all_x, all_y):
            error = model.update(x, y)
            error_count += abs(error)

        print(f"Epoch {epoch+1} errors {error_count}")
```

```python
ppn = Perceptron(num_features=2)

train(model=ppn, all_x=X_train, all_y=y_train, epochs=5)
```

```python
def compute_accuracy(model, all_x, all_y):

    correct = 0.0

    for x, y in zip(all_x, all_y):
        prediction = model.forward(x)
        correct += int(prediction == y)

    return correct / len(all_y)
```

```python
train_acc = compute_accuracy(ppn, X_train, y_train)
train_acc
```
