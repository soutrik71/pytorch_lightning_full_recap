In this notebook we will focus on fundamentals of pytorch and tensor operations.

```python
import torch
import matplotlib.pyplot as plt
```

```python
# Rank-0 tensor
a = torch.tensor(1)
print(a)
print(a.shape)
print(a.ndim)
```

```python
# rank-1 tensor
b = torch.tensor([1,2,3,4])
print(b)
print(b.shape)
print(b.ndim)
print(b.dtype)
# changing datatype of tensor
b = b.to(torch.int32)
print(b)
print(b.dtype)
```

```python
# rank-2 tensor 
c = torch.tensor([[1.2,2.1,3.6],
                    [4.5,5.9,6.7],
                    [7.8,8.3,9.0]])

print(c)
print(c.ndim)
print(c.shape)
print(c.dtype)
print(c.device)
# changing datatype of tensor
c = c.to(torch.float16)
print(c)
print(c.dtype)
```
